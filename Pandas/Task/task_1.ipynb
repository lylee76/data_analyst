```python
#create table using dictionary

import pandas as pd

data = {'Employee':['John','Kate','Leo'],
        'Department':['IT','HR','Sales'],
        'Salary':[50000,55000,60000]}

df = pd.DataFrame(data)
print(df)
```

```python
#create table using list of list

data = [['John','Kate','Leo'],
        ['IT','HR','Sales'],
        [50000,55000,60000]]

df= pd.DataFrame(data, columns=['Employee', 'Department', 'Salary'])
print(df)
```
